```python
import pandas as pd

df = pd.read_csv('student_dataset.csv')
df.head()
```

```python
#set student_id as index
df_idx = df.set_index('student_id')
df_idx.head()
```

```python
#.loc lookup method by student_id
some_id = df['student_id'].iloc[0]
df_idx.loc[some_id]
```

```python
#turn student_id back into regular column
df_reset = df_idx.reset_index()
df_reset.head()
```

```python
#set hierarchical index
df_mi = df.set_index(['major', 'grad_year'])
df_mi.head()
```

```python
#sort the multi index
df_mi_sorted = df_mi.sort_index()
df_mi_sorted.head()
```

```python
#select rows for CS major and 2025 graduation year
df_mi_sorted.loc[('CS', 2025)]
```

```python
#boolean lookup method
some_id = df["student_id"].iloc[len(df)//2]
df[df["student_id"] == some_id]
```

```python
#index lookup method
df_idx = df.set_index("student_id")
df_idx.loc[some_id]
```

```python
#time boolean method
%%timeit
_ = df[df["student_id"] == some_id]
```

```python
#time lookup method
%%timeit
_ = df_idx.loc[some_id]
```

# Reflection

It would appear that the index lookup method runs much faster. This is because the boolean method has to scan the entire csv, while the index method uses the index for quick, direct access.
Resetting an index is helpful when exploration of a dataset involves observing the data from mulitple perspectives in which numerous columns might need to be considered as the index. Being able to reset an index easily allows for efficient exploration with diverse lookup function performance.
Indexing in both Pandas and SQL both utilize time effifiency by not scanning every row of a database, which makes dealing with larger systems far more efficient.
