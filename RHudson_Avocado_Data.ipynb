```python
import sqlite3

KAGGLE_CSV = "avocado.csv"
NY_JSON    = "avocado_secondary_NY.json"
SF_EXCEL   = "avocado_secondary_SF.xlsx"
SQLITE_DB  = "avocado_secondary_ALL.db"
```

```python
df_kaggle = pd.read_csv(KAGGLE_CSV)

display(df_kaggle.head())
print("Kaggle shape:", df_kaggle.shape)
df_kaggle.info()
```

```python
kaggle_cols = list(df_kaggle.columns)
print("Kaggle columns:", kaggle_cols)

#columns I must match
```

```python
#load and inspect NY JSON
df_ny = pd.read_json(NY_JSON, lines=True)

display(df_ny.head())
print("NY shape:", df_ny.shape)
df_ny.info()

#Convert epoch milliseconds to datetime
df_ny["Date"] = pd.to_datetime(df_ny["Date"], unit="ms", errors="coerce")
```

```python
def align_to_kaggle_schema(df_secondary: pd.DataFrame, kaggle_columns: list) -> pd.DataFrame:
    df = df_secondary.copy()

    #Standardize column names
    rename_map = {
        "TotalVolume": "Total Volume",
        "Total_Volume": "Total Volume",
        "avg_price": "AveragePrice",
        "average_price": "AveragePrice",
        "Region": "region",
        "Type": "type",
    }
    df = df.rename(columns={c: rename_map.get(c, c) for c in df.columns})

    #Standardize region/type formatting
    if "region" in df.columns:
        df["region"] = df["region"].astype(str).str.strip()
    if "type" in df.columns:
        df["type"] = df["type"].astype(str).str.strip().str.lower()

    #Ensure required core fields exist
    required = ["Date", "AveragePrice", "Total Volume", "type", "region"]
    for col in required:
        if col not in df.columns:
            df[col] = pd.NA

    #Fix dtypes for core numeric fields
    df["AveragePrice"] = pd.to_numeric(df["AveragePrice"], errors="coerce")
    df["Total Volume"] = pd.to_numeric(df["Total Volume"], errors="coerce")

    #Add time fields
    df["year"] = df["Date"].dt.year

    #Add any Kaggle columns missing in secondary data
    for col in kaggle_columns:
        if col not in df.columns:
            df[col] = pd.NA

    #Reorder columns to match Kaggle
    df = df[kaggle_columns]

    return df

df_ny_aligned = align_to_kaggle_schema(df_ny, kaggle_cols)
df_sf_aligned = align_to_kaggle_schema(df_sf, kaggle_cols)

print("NY aligned shape:", df_ny_aligned.shape)
print("SF aligned shape:", df_sf_aligned.shape)

display(df_ny_aligned.head())
display(df_sf_aligned.head())
```

```python
#handle any missing

def clean_secondary(df_aligned: pd.DataFrame) -> pd.DataFrame:
    df = df_aligned.copy()

    #Drop rows missing Date, AveragePrice, or Total Volume (critical)
    df = df.dropna(subset=["Date", "AveragePrice", "Total Volume", "type", "region"])

    #Remove impossible values
    df = df[df["AveragePrice"] > 0]
    df = df[df["Total Volume"] >= 0]

    return df
```

```python
df_ny_clean = clean_secondary(df_ny_aligned)
df_sf_clean = clean_secondary(df_sf_aligned)

print("NY clean shape:", df_ny_clean.shape)
print("SF clean shape:", df_sf_clean.shape)
```

```python
#merge NY and SF into one

df_secondary_merged = pd.concat([df_ny_clean, df_sf_clean], ignore_index=True)

print("Merged secondary shape:", df_secondary_merged.shape)
display(df_secondary_merged.head())
```

```python
#SQL Validation
conn = sqlite3.connect(SQLITE_DB)

#Read avocado_all table from SQLite
df_sqlite = pd.read_sql_query("SELECT * FROM avocado_all;", conn)

conn.close()
```

```python
print("SQLite shape:", df_sqlite.shape)
display(df_sqlite.head())
df_sqlite.info()
```

```python
#Validation checks cont

#column alignment
secondary_cols = set(df_secondary_merged.columns)
sqlite_cols = set(df_sqlite.columns)
```

```python
print("Columns only in secondary merged:", sorted(list(secondary_cols - sqlite_cols)))
print("Columns only in SQLite:", sorted(list(sqlite_cols - secondary_cols)))
print("Shared columns count:", len(secondary_cols & sqlite_cols))
```

```python
#row counts
print("Row count - secondary merged:", len(df_secondary_merged))
print("Row count - SQLite:", len(df_sqlite))
```

```python
#data type comparisons
shared_cols = sorted(list(secondary_cols & sqlite_cols))
dtype_compare = pd.DataFrame({
    "column": shared_cols,
    "secondary_dtype": [df_secondary_merged[c].dtype for c in shared_cols],
    "sqlite_dtype": [df_sqlite[c].dtype for c in shared_cols],
})
display(dtype_compare)
```

```python
#Comparing overlapping rows

#datetime in both
df_secondary_merged["Date"] = pd.to_datetime(df_secondary_merged["Date"], errors="coerce")

#parsing text or numeric
df_sqlite["Date"] = pd.to_datetime(df_sqlite["Date"], errors="coerce")

#compare columns
compare_cols = ["Date", "region", "type", "AveragePrice", "Total Volume"]

sec_small = df_secondary_merged[compare_cols].copy()
sql_small = df_sqlite[compare_cols].copy()
```

```python
#Inner join to find overlapping keys
overlap = sec_small.merge(
    sql_small,
    on=["Date", "region", "type"],
    how="inner",
    suffixes=("_sec", "_sql")
)

print("Overlapping key rows found:", overlap.shape[0])
display(overlap.head())
```

```python
#Compare numeric differences
overlap["price_diff"] = (overlap["AveragePrice_sec"] - overlap["AveragePrice_sql"]).abs()
overlap["volume_diff"] = (overlap["Total Volume_sec"] - overlap["Total Volume_sql"]).abs()

print("Price diff summary:")
display(overlap["price_diff"].describe())

print("Volume diff summary:")
display(overlap["volume_diff"].describe())
```

```python
#How many match
price_tol = 0.001
volume_tol = 0.01
print("Rows within price tolerance:", (overlap["price_diff"] <= price_tol).sum())
print("Rows within volume tolerance:", (overlap["volume_diff"] <= volume_tol).sum())
```
